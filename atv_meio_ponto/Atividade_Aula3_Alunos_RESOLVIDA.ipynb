```python
import pandas as pd
```

```python
# Coloque o arquivo "Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
# na MESMA PASTA deste notebook para o caminho relativo abaixo funcionar.
# Se preferir, troque pelo caminho completo do arquivo no seu computador, ex:
# path = r"C:\Users\SeuUsuario\Documents\Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
path = "Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
```

```python
formacao_bruto = pd.read_excel(path, header=None)
```

```python
formacao_bruto
```

A tabela deverá ser dividida em 3
- homens e mulheres por região com curso superior
- homens e mulheres por região com curso nas áreas de Ciência, Tecnologia, Engenharias e Matemática
- homens e mulheres por região com curso nas áreas de Educação, Serviços pessoais, Saúde e Bem-estar	

Exemplo 1 — Curso superior (todas as áreas)

```python
# : = todas as linhas  |  0:4 = colunas 0, 1, 2 e 3 (as 4 primeiras)
formacao_total = formacao_bruto.iloc[3:36, 0:4].copy()
```

```python
formacao_total.columns = [
    "uf_regiao",
    "total",
    "total_homens",
    "total_mulheres"
]
```

```python
formacao_total
```

```python
formacao_total["percentual_mulheres"] = ((formacao_total['total_mulheres'] / formacao_total['total']) * 100).round(2)
```

```python
formacao_total
```

**Leitura rápida do Exemplo 1:** no Brasil, mulheres já são **59,47%** dos formados no ensino superior (quase 6 em cada 10). Isso é maioria em todas as regiões — o Nordeste é onde essa maioria é mais forte (63,26%) e o Sudeste onde é mais fraca (57,97%), mas em nenhum lugar os homens ultrapassam as mulheres.

## Exemplo 2 — Ciência, Tecnologia, Engenharias e Matemática (STEM)

```python
# : = todas as linhas | as colunas dessa área ficam nas posições 0 (uf/região), 4, 5 e 6
formacao_stem = formacao_bruto.iloc[3:36, [0, 4, 5, 6]].copy()
```

```python
formacao_stem.columns = [
    "uf_regiao",
    "total",
    "total_homens",
    "total_mulheres"
]
```

```python
formacao_stem
```

```python
formacao_stem["percentual_mulheres"] = ((formacao_stem['total_mulheres'] / formacao_stem['total']) * 100).round(2)
```

```python
formacao_stem
```

**Leitura rápida do Exemplo 2:** em Ciência, Tecnologia, Engenharias e Matemática o quadro se inverte — as mulheres caem para **33,71%** dos formados no Brasil, ou seja, são minoria em todas as regiões. O Sudeste tem a menor participação feminina (31,75%) e o Nordeste a maior (37,24%), mas nenhuma região chega perto dos 50%.

## Exemplo 3 — Educação, Serviços pessoais, Saúde e Bem-estar

```python
# : = todas as linhas | as colunas dessa área ficam nas posições 0 (uf/região), 7, 8 e 9
formacao_saude = formacao_bruto.iloc[3:36, [0, 7, 8, 9]].copy()
```

```python
formacao_saude.columns = [
    "uf_regiao",
    "total",
    "total_homens",
    "total_mulheres"
]
```

```python
formacao_saude
```

```python
formacao_saude["percentual_mulheres"] = ((formacao_saude['total_mulheres'] / formacao_saude['total']) * 100).round(2)
```

```python
formacao_saude
```

**Leitura rápida do Exemplo 3:** em Educação, Serviços pessoais, Saúde e Bem-estar as mulheres são amplamente maioria: **79,23%** dos formados no Brasil, chegando a **80,21%** no Sul e **80,03%** no Nordeste — as duas maiores concentrações do país.

## Comparando as três áreas

```python
# Junta o "% de mulheres" das três tabelas em uma só, por região, para comparar lado a lado
comparacao = formacao_total[["uf_regiao", "percentual_mulheres"]].rename(
    columns={"percentual_mulheres": "pct_mulheres_total"}
)
comparacao = comparacao.merge(
    formacao_stem[["uf_regiao", "percentual_mulheres"]].rename(
        columns={"percentual_mulheres": "pct_mulheres_stem"}
    ),
    on="uf_regiao"
)
comparacao = comparacao.merge(
    formacao_saude[["uf_regiao", "percentual_mulheres"]].rename(
        columns={"percentual_mulheres": "pct_mulheres_saude"}
    ),
    on="uf_regiao"
)

# Só as grandes regiões + Brasil, para facilitar a leitura
comparacao_regioes = comparacao[comparacao["uf_regiao"].isin(
    ["Brasil", "Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
)].reset_index(drop=True)

comparacao_regioes
```

```python
brasil = comparacao.loc[comparacao["uf_regiao"] == "Brasil"].iloc[0]

print("Brasil — % de mulheres entre os formados, por área de curso")print(f"  Curso superior (geral):                     {brasil.pct_mulheres_total}%")print(f"  Ciência, Tecnologia, Eng. e Matemática:      {brasil.pct_mulheres_stem}%")print(f"  Educação, Serviços pessoais e Saúde:         {brasil.pct_mulheres_saude}%")
```

**Conclusão:** no ensino superior como um todo as mulheres já são maioria (quase 6 em cada 10 formados). Mas essa maioria não se distribui igualmente entre as áreas: elas são **minoria** em Ciência, Tecnologia, Engenharias e Matemática (1 em cada 3) e **super-representadas** em Educação, Serviços pessoais e Saúde (quase 8 em cada 10) — um padrão de segregação de gênero por área de formação que se repete em todas as regiões do país.
